## **Explicación del Código de Ejemplo**

Este código es un ejemplo de bloques lógicos anidados que funcionan como un filtro.

El objetivo es crear una lista (resultado) solo con las parejas de números (x, y) que cumplan una condición específica.

Para lograrlo, el código revisa sistemáticamente todas las combinaciones posibles: primero toma un número x (del 0 al 9) y luego, para ese x, lo combina con cada número y (del 0 al 4).

A cada pareja le aplica una regla: si la multiplicación x * y es mayor o igual a 10, la pareja pasa el filtro y se guarda en la lista final. Al final, se imprime la lista con todas las parejas que cumplieron la condición.

```python
resultado = []
for x in range(10):
    for y in range(5):
        if x * y >= 10:
            resultado.append((x, y))
print(resultado)
```

## **Notación Big-O y orden de rapidez**

La notación Big-O describe cómo crece el tiempo de ejecución (o uso de memoria) de un algoritmo cuando aumenta el tamaño de entrada
𝑛
n.
En general, mientras más lento crece la función, más rápido será el algoritmo para entradas grandes.

Orden común de complejidades (de más rápida a más lenta):

O(1) – Tiempo constante: no importa el tamaño de entrada (ej. acceso a un elemento en una lista).

O(log n) – Logarítmica: crece muy lentamente (ej. búsqueda binaria).

O(n) – Lineal: el tiempo crece proporcionalmente al tamaño (ej. recorrer una lista).

O(n log n) – Lineal-logarítmica: típico de algoritmos de ordenamiento eficientes (ej. mergesort).

O(n²) – Cuadrática: se vuelve lenta con listas grandes (ej. burbuja, inserción simple).

O(2ⁿ) – Exponencial: se vuelve impráctica muy rápido (ej. fuerza bruta en problemas combinatorios).

O(n!) – Factorial: extremadamente lenta incluso para valores pequeños (ej. permutaciones completas).

```python
# Big-O demo con 7 complejidades y animación simple en Colab/Jupyter
# Reglas cumplidas: matplotlib, una sola gráfica, sin estilos/colores específicos.

import time
import math
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output, display

# --------- Funciones representativas ---------
def op_constante(n):
    # O(1): hace trabajo constante, independiente de n
    a = 42
    a += 1
    a *= 2
    return a

def op_log(n):
    # O(log n): duplica hasta pasar n
    x = 1
    while x < n:
        x *= 2

def op_lineal(n):
    # O(n): recorre n pasos
    for _ in range(n):
        pass

def op_nlogn(n):
    # O(n log n): n veces un trabajo logarítmico
    for _ in range(n):
        x = 1
        while x < n:
            x *= 2

def op_cuadratica(n):
    # O(n^2): dos bucles anidados
    for _ in range(n):
        for _ in range(n):
            pass

def op_exponencial(n):
    # O(2^n): simular 2**n "unidades de trabajo"
    # n pequeños/medianos para no bloquear
    for _ in range(2 ** n):
        pass

def op_factorial(n):
    # O(n!): simular n! "unidades de trabajo" (n muy pequeño)
    total = math.factorial(n)
    for _ in range(total):
        pass

# --------- Configuración de tamaños ---------
# Elegidos para que la animación sea perceptible pero breve
experimentos = [
    ("O(1)", op_constante, [10, 10**3, 10**6, 10**9]),
    ("O(log n)", op_log, [10**3, 10**4, 10**5, 10**6, 10**7]),
    ("O(n)", op_lineal, [2_000, 8_000, 20_000, 40_000, 80_000]),
    ("O(n log n)", op_nlogn, [300, 800, 1_600, 3_200, 6_400]),
    ("O(n^2)", op_cuadratica, [60, 100, 140, 180, 220]),
    ("O(2^n)", op_exponencial, [12, 14, 16, 18, 20, 22]),     # 2^22 ≈ 4.2M
    ("O(n!)", op_factorial, [7, 8, 9, 10])                    # 10! ≈ 3.6M
]

# --------- Estructuras para resultados ---------
resultados = {nombre: {"n": [], "t": []} for nombre, _, _ in experimentos}

# --------- Crear la figura (una sola) ---------
fig, ax = plt.subplots(figsize=(8, 5))
ax.set_title("Crecimiento temporal (Big-O) — 7 complejidades (animado)")
ax.set_xlabel("Tamaño de entrada n")
ax.set_ylabel("Tiempo (segundos)")
ax.set_xscale("log")
ax.set_yscale("log")
ax.grid(True)
plt.tight_layout()

# --------- Animación paso a paso ---------
for nombre, fn, ns in experimentos:
    for n in ns:
        t0 = time.perf_counter()
        fn(n)
        t1 = time.perf_counter()
        dt = t1 - t0

        resultados[nombre]["n"].append(n)
        resultados[nombre]["t"].append(dt)

        # Redibujar
        clear_output(wait=True)
        ax.cla()
        ax.set_title("Crecimiento temporal (Big-O) — 7 complejidades (animado)")
        ax.set_xlabel("Tamaño de entrada n")
        ax.set_ylabel("Tiempo (segundos)")
        ax.set_xscale("log")
        ax.set_yscale("log")
        ax.grid(True)

        # Dibujar cada serie disponible
        for serie_nombre, data in resultados.items():
            if data["n"]:
                ax.plot(data["n"], data["t"], marker="o", label=serie_nombre)

        ax.legend(loc="best")
        ax.set_ylim(bottom=1e-6)

        display(fig)
        print(f"Registrado {nombre} con n={n} → {dt:.6f} s")

        # Pausa corta para visualizar la progresión
        time.sleep(0.35)

# Nota: si alguna serie tarda demasiado en tu equipo, reduce los tamaños de 'n' correspondientes.

```

### **Construir un data frame (CSV) de ejemplo**

En esta sección, vamos a "inventar" un conjunto de datos para la clase usando las librerías numpy y pandas. A esta tabla la llamaremos da.

El código crea una tabla (DataFrame) con 250 filas y le añade columnas con datos generados aleatoriamente que simulan una encuesta de salud. Además, calcula el Índice de Masa Corporal (IMC) y agrega intencionalmente algunos valores vacíos para practicar cómo se manejan los datos incompletos en un escenario real.

Cada variable (columna) representa lo siguiente:

*  SEQN: Un identificador único para cada persona en el estudio.

*  RIAGENDR: Género, representado con números (1 y 2). Es una variable categórica.

*  RIDAGEYR: Edad de la persona en años.

*  BMXWT: Peso en kilogramos.

*  BMXHT: Altura en centímetros.

*  BMXBMI: Índice de Masa Corporal (IMC), que se calcula a partir del peso y la altura.

*  PAIS: País de origen de la persona. Esta es otra variable categórica que hemos añadido.

```python
#Cargada una vez la libreria, no se debe volver a cargar de nuevo.
import numpy as np
import pandas as pd

# Fijar una semilla para que los resultados aleatorios sean siempre los mismos
np.random.seed(42)

# Número de filas que tendrá nuestra tabla
n = 250

# Crear el DataFrame
da = pd.DataFrame({
    "SEQN": np.arange(1, n+1),
    "RIAGENDR": np.random.choice([1, 2], size=n, p=[0.5, 0.5]), # 1: Masculino, 2: Femenino
    "RIDAGEYR": np.random.randint(18, 80, size=n),
    "BMXWT": np.round(np.random.normal(70, 12, size=n), 1), # Peso en kg
    "BMXHT": np.round(np.random.normal(168, 10, size=n), 1), # Altura en cm
    "PAIS": np.random.choice(["Colombia", "Mexico", "Argentina"], size=n, p=[0.6, 0.2, 0.2]) # Variable categórica de texto
})

# Calcular el IMC y añadirlo como una nueva columna
da["BMXBMI"] = np.round(da["BMXWT"] / (da["BMXHT"]/100)**2, 1)

# Simular datos faltantes: seleccionar 10 filas al azar y poner NaN en la columna de peso
da.loc[np.random.choice(da.index, size=10, replace=False), "BMXWT"] = np.nan

# Mostrar las primeras 5 filas de nuestra tabla para revisarla
da.head()
```

### **Slicing: Seleccionar Filas y Columnas**

"Slicing" (o rebanar) es la técnica que usamos para seleccionar pedazos específicos de nuestra tabla. Podemos querer ver solo una columna (como la edad de todos los participantes) o una fila completa (todos los datos de una persona específica).

Pandas nos ofrece varias formas de hacerlo, siendo las más comunes .loc (para seleccionar por nombre) e .iloc (para seleccionar por número de posición).

* **Seleccionar una Columna**
Hay múltiples maneras de seleccionar una sola columna. Aunque la sintaxis varía, todas las opciones que se muestran a continuación logran exactamente lo mismo: aislar la columna RIDAGEYR (edad).

* **Seleccionar una Fila**
Para seleccionar una fila por su posición, usamos .iloc. Es importante recordar que en Python se empieza a contar desde 0, por lo que .iloc[3] en realidad selecciona la cuarta fila de la tabla.

```python
# --- Formas de seleccionar la columna de edad (RIDAGEYR) ---

# Opción 1: Usando corchetes y el nombre de la columna (la más común)
w = da["RIDAGEYR"]

# Opción 2: Usando .loc para seleccionar por nombre
x = da.loc[:, "RIDAGEYR"] # El ":" significa "todas las filas"

# Opción 3: Usando un punto (solo funciona si el nombre no tiene espacios)
y = da.RIDAGEYR

# Opción 4: Usando .iloc para seleccionar por número de columna
z = da.iloc[:, da.columns.get_loc("RIDAGEYR")]

# Verificamos que todas las formas dieron el mismo resultado
print("Max RIDAGEYR (cuatro formas):", w.max(), x.max(), y.max(), z.max())


# --- Seleccionar la cuarta fila de la tabla (índice 3) ---
fila4 = da.iloc[3, :] # El ":" al final significa "todas las columnas"
display(fila4)
```

### **Manejo de Datos Faltantes (NaN)**

En los conjuntos de datos del mundo real, es muy común que falte información. Pandas representa estos vacíos con un valor especial llamado NaN (Not a Number). Antes de hacer cualquier análisis, es crucial identificar cuántos datos faltantes tenemos y decidir qué hacer con ellos.


**¿Qué hacer con los datos faltantes?**
Una vez identificados, existen varias estrategias comunes para manejarlos. La elección depende del problema y de la cantidad de datos que falten.

**Eliminar las filas:** Si tienes muy pocos datos faltantes en un conjunto de datos muy grande, la solución más simple es eliminar las filas completas que contienen un NaN.

  Ventaja: Es rápido y fácil.

  Desventaja: Pierdes la información de las otras columnas de esa misma fila.

**Rellenar (imputar) los valores:** Consiste en reemplazar el NaN con un valor calculado a partir del resto de la columna.

* **Opción A (para números):** Rellenar con el promedio (media) o la mediana de la columna. La mediana suele ser mejor si hay valores atípicos.

* **Opción B (para categorías):** Rellenar con la moda (el valor que más se repite).

  Ventaja: No pierdes filas enteras de datos.

  Desventaja: Puede introducir un pequeño sesgo en los datos.

**Dejarlos como están**: Algunos modelos de machine learning son capaces de manejar los valores faltantes por sí mismos, aunque no es lo más común.




El siguiente código cuenta cuántos valores faltantes (NaN) y no faltantes hay en la columna de peso (BMXWT).

```python
faltantes_bmxwt = pd.isnull(da["BMXWT"]).sum()
no_faltantes_bmxwt = pd.notnull(da["BMXWT"]).sum()

print("BMXWT → faltantes:", faltantes_bmxwt, "| no faltantes:", no_faltantes_bmxwt)
```

### **Visualizar una Variable Numérica (Histograma)**

Para las variables numéricas como la edad (RIDAGEYR), el histograma es el gráfico ideal. Agrupa los números en "cajones" o intervalos llamados bins y nos muestra cuántos datos caen en cada uno.

Para decidir cuántos bins usar, una regla clásica es la Regla de Sturges, que busca un balance para que el gráfico sea legible. La fórmula es k = 1 + 3.322 * log10(n), donde n es el número de datos. Existen otras reglas como la de Scott o Freedman-Diaconis que también son muy útiles.

```python
# --- Histograma para la Edad (RIDAGEYR) ---

# 1. Calcular el número de bins usando la Regla de Sturges
n = len(da)
k = int(1 + 3.322 * np.log10(n))
print(f"Número de datos (n): {n}")
print(f"Bins sugeridos por Sturges (k): {k}")


# 2. Crear el histograma
plt.figure(figsize=(10, 5))
plt.hist(da["RIDAGEYR"].dropna(), bins=k, color='skyblue', edgecolor='black')

# Añadir títulos y etiquetas
plt.title('Distribución de Edades (RIDAGEYR)')
plt.xlabel('Edad (años)')
plt.ylabel('Frecuencia (cantidad de personas)')
plt.grid(axis='y', alpha=0.75)

# Mostrar el gráfico
plt.show()
```

### **Visualizar una Variable Categórica (Gráfico de Barras)**
Para variables categóricas como el país (PAIS), donde los datos son etiquetas de texto, el gráfico de barras es perfecto. Simplemente cuenta cuántas veces aparece cada categoría y muestra el total con una barra.

```python
# --- Gráfico de Barras para el País (PAIS) ---

# 1. Contar la frecuencia de cada categoría
conteo_paises = da['PAIS'].value_counts()
print("Conteo por país:")
print(conteo_paises)


# 2. Crear el gráfico de barras
plt.figure(figsize=(8, 5))
plt.bar(conteo_paises.index, conteo_paises.values, color=['#0033A0', '#FCD116', '#CE1126'])

# Añadir títulos y etiquetas
plt.title('Distribución de Personas por País')
plt.xlabel('País')
plt.ylabel('Cantidad de Personas')

# Mostrar el gráfico
plt.show()
```
